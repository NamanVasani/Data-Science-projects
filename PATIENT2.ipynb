```python
patient["risk_level_num"] = patient["risk_level"].map({
    "Low": 1,
    "Medium": 2,
    "High": 3
})

# Create figure
plt.figure(figsize=(8,6))

# Scatter plot
scatter = plt.scatter(
    patient["age"],
    patient["hospital_bill_usd"],
    c=patient["risk_level_num"],
    cmap="viridis",
    s=100,
    alpha=0.8,
    edgecolors="black"
)

# Labels and title
plt.xlabel("Age", fontsize=12)
plt.ylabel("Hospital Bill USD", fontsize=12)
plt.title("Age vs Hospital Bill based on Risk Level", fontsize=14)

# Color bar
cbar = plt.colorbar(scatter)
cbar.set_label("Risk Level")

# Grid
plt.grid(True)

# Show plot
plt.show()
```

```python
patient["smoking"]=patient["smoking"].map({
    "Yes":2,
    "Former":1,
    "No":0
})
plt.figure(figsize=(5,6))
scatter1=plt.scatter(
    patient["age"],
    patient["hospital_bill_usd"],
    c=patient["smoking"],
    cmap="inferno",
    s=100,
    alpha=0.9,
    edgecolors="red"
)
plt.title(" Age VS Hospital_bill_usd by Smoking")
plt.xlabel("AGE")
plt.ylabel("hospital_bill_usd")
cbar=plt.colorbar(scatter1)
cbar.set_label("Smoking")
plt.grid(True)
plt.show()
```

```python


# Chart
plt.figure(figsize=(10,6))
scatter1 = plt.scatter(
    patient["bmi"],
    patient["glucose_level"],
    c=patient["smoking"],
    cmap="inferno",
    s=100,
    alpha=0.9,
    edgecolors="black"
)
cbar = plt.colorbar(scatter1)
cbar.set_label("Smoking (0=No, 1=Former, 2=Yes)")
plt.title("BMI vs Glucose Level by Smoking")
plt.xlabel("BMI")
plt.ylabel("Glucose Level")
plt.grid(True)
plt.show()
```

```python
# All numerical columns in one figure
fig, axes = plt.subplots(2,4,figsize=(16,8))
cols = ['age', 'bmi', 'bp_systolic', 'bp_diastolic',
        'heart_rate', 'glucose_level', 'hba1c', 'hospital_bill_usd']

for i, col in enumerate(cols):
    ax = axes[i//4][i%4]
    ax.boxplot(patient[col].dropna())
    ax.set_title(col)

plt.suptitle('Outlier Analysis — Patient')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(
    patient.corr(numeric_only=True),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)
plt.title('Correlation Heatmap — Patient')
plt.show()
```
